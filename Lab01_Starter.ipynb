### C2 — โค้ดที่จัดให้ — อ่าน รัน และตรวจผล ไม่ต้องเขียนฟังก์ชันนี้เอง

เขียนผลคาดการณ์ในช่องคำตอบก่อนรันกรณีทดลองที่เกี่ยวข้อง

````python
def inspect_orders(customers, orders):
    checked = orders.copy(deep=True)
    dates = pd.to_datetime(checked["order_date"], format="%Y-%m-%d", errors="coerce")
    issues = []
    rules = [
        (checked["order_id"].isna() | checked["order_id"].astype(str).str.strip().eq(""), "missing_order_id"),
        (checked.duplicated("order_id", keep=False), "duplicate_order_id"),
        (~checked["customer_id"].isin(customers["customer_id"]), "customer_not_found"),
        (dates.isna(), "invalid_order_date"),
    ]
    for mask, reason in rules:
        for row_number in checked.index[mask]:
            issues.append({"row_index": row_number, "order_id": checked.loc[row_number, "order_id"], "reason": reason})
    checked["order_date"] = dates
    return checked, pd.DataFrame(issues, columns=["row_index", "order_id", "reason"])

def build_summary(customers, orders, cutoff=CUTOFF, as_of=AS_OF):
    checked, issues = inspect_orders(customers, orders)
    if not issues.empty:
        return None, pd.DataFrame(), issues
    in_window = checked["order_date"].between(HISTORY_START, cutoff, inclusive="both")
    history = checked.loc[in_window].copy()
    excluded = checked.loc[~in_window].copy()
    last = history.groupby("customer_id", as_index=False)["order_date"].max()
    last = last.rename(columns={"order_date": "last_order_date"})
    summary = customers[["customer_id"]].merge(last, on="customer_id", how="left", validate="one_to_one")
    summary["days_since_last_order"] = (as_of - summary["last_order_date"]).dt.days.astype("Int64")
    summary["as_of_date"] = as_of
    return summary.sort_values("customer_id").reset_index(drop=True), excluded, issues

baseline_summary, baseline_excluded, baseline_issues = build_summary(raw_customers, raw_orders)
print("Baseline", DATA_VERSION, "history_start:", HISTORY_START.date(), "cutoff:", CUTOFF.date(), "as_of:", AS_OF.date())
display(baseline_summary)
history_view = raw_orders.sort_values(["customer_id", "order_date", "order_id"]).copy()
history_view["gap_days"] = history_view.groupby("customer_id")["order_date"].diff().dt.days.astype("Int64")
display(history_view)
print("Excluded:", len(baseline_excluded), "Issues:", len(baseline_issues))
````
